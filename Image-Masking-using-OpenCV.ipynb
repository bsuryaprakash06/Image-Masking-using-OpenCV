```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
yellow_square = cv2.imread("yellow_square.png")
green_circle = cv2.imread("green_circle.png")

yellow_square = cv2.cvtColor(yellow_square, cv2.COLOR_BGR2RGB)
green_circle = cv2.cvtColor(green_circle, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(yellow_square)
plt.title("Yellow Square")

plt.subplot(122)
plt.imshow(green_circle)
plt.title("Green Circle")

plt.show()
```

## Question 1

Convert the **green circle image** into **grayscale** using OpenCV.

Use:

cv2.cvtColor()

Store the result in the variable:

img_gray_circle

```python
img_gray_circle = cv2.cvtColor(green_circle, cv2.COLOR_RGB2GRAY)
```

## Question 2

Create a **binary mask** that isolates the **ring region**.

Use:

cv2.threshold()

Hint:
Choose a threshold value that captures the **light ring**.

Store the mask in:

img_ring_mask

```python
retval, img_ring_mask = cv2.threshold(img_gray_circle, 10, 255, cv2.THRESH_BINARY)

plt.imshow(img_ring_mask, cmap='gray')
plt.show()
```

## Question 3

Create an **inverse mask** using:

cv2.bitwise_not()

Store the result in:

img_ring_mask_inv

```python
img_ring_mask_inv = cv2.bitwise_not(img_ring_mask)
```

## Question 4

Use **bitwise operations** to combine the **inverse mask**
with the **yellow square image**.

The final output should show:

- Yellow background
- Black ring

Store the result in:

img_combined

```python
img_ring_mask = cv2.resize(img_ring_mask,(yellow_square.shape[1], yellow_square.shape[0]))
img_ring_mask_inv = cv2.bitwise_not(img_ring_mask)

img_combined = cv2.bitwise_and(yellow_square, yellow_square, mask = img_ring_mask_inv)
```

```python
plt.figure(figsize=(20,10))

plt.subplot(141)
plt.imshow(img_gray_circle, cmap='gray')
plt.title("Gray Scale Image")

plt.subplot(142)
plt.imshow(img_ring_mask, cmap='gray')
plt.title("Mask")

plt.subplot(143)
plt.imshow(img_ring_mask_inv, cmap='gray')
plt.title("Inverse Mask")

plt.subplot(144)
plt.imshow(img_combined)
plt.title("Combined Result")

plt.show()
```
